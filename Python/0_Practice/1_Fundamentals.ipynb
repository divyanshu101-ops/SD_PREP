#### Python is dynamically typed, meaning the type associated with a variable is determined during runtime, and a variable can refer to objects of different types at different times.

```python
# Python was designed this way to keep coding simple, consistent, and flexible, sparing developers from juggling 
# separate rules for primitive types and objects. Its object-oriented nature is rooted in the fact 
# that every single entity—from a basic number or string to an entire function—is created in memory as an 
# instance (object) of a class. When you assign x = 10, it is not just a raw memory value, 
# but an instance of the int class equipped with its own attributes and methods, 
# where an operation like 10 + 5 actually executes the method call (10).__add__(5) under the hood. 
# Because of this pure object model, functions can be passed around just like ordinary variables, 
# and custom classes can interact with the language just as seamlessly as any built-in data type.
```

```python
# for Loop

# Python's for loop commonly iterates over an iterable.

for i in range(5):
    print(i)

# Why 5 isn't included?

# Because:
# range(5)

# means:
# 0 ≤ i < 5
```

```python
i = 0

while i < 5:
    print(i)
    i += 1
```

```python
for i in range(10):
    if i == 5:
        break

    print(i)
```

```python
for i in range(5):
    if i == 2:
        continue

    print(i)
```
