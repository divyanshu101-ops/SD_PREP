# LIST

```python
# List
# │
# ├── Ordered
# ├── Indexable
# ├── Sliceable
# ├── Mutable
# ├── Allows duplicates
# ├── Can contain different types
# │
# ├── append()
# ├── extend()
# ├── insert()
# ├── remove()
# ├── pop()
# ├── clear()
# ├── index()
# ├── count()
# ├── sort()
# └── reverse()
```

```python
# What is a List?
# A list is an ordered, mutable collection.
```

```python
numbers = [10, 20, 30, 40]
# A list can contain different types:
data = [10, "hello", 3.14, True]
# Python allows this.
```

```python
# Lists are Mutable

# This is the biggest difference from strings.

numbers = [10, 20, 30]

numbers[0] = 100

print(numbers)
```

```python
# List Slicing
# Same syntax as strings:

numbers = [10, 20, 30, 40, 50]

numbers[1:4]
numbers[::-1]
```

```python
# append()

# Adds one element at the end.

numbers = [1, 2, 3]

numbers.append(4)

print(numbers)
```

```python
# extend()

# Adds multiple elements.

numbers = [1, 2]

numbers.extend([3, 4, 5])

print(numbers)
```

```python
# append() vs extend()
# This is a very common interview question.

a = [1, 2]

a.append([3, 4])

# Result:
[1, 2, [3, 4]]

# But:

a = [1, 2]

a.extend([3, 4])

# Result:

[1, 2, 3, 4]

# append(x)
# → add x as ONE element

# extend(iterable)
# → add elements from iterable individually
```

```python
# insert()

# Adds an element at a particular index.

numbers = [10, 20, 30]

numbers.insert(1, 15)

print(numbers)
```

```python
# remove()

# Removes the first matching value.

numbers = [10, 20, 30, 20]

numbers.remove(20)

print(numbers)
```

```python
# pop()

# Removes and returns an element.

numbers = [10, 20, 30]

x = numbers.pop()

print(x)
print(numbers)
```

```python
# clear()
# Removes everything.

numbers = [1, 2, 3]

numbers.clear()

print(numbers)
```

```python
# index()

# Returns index of first occurrence.
numbers = [10, 20, 30]

print(numbers.index(20))
```

```python
# count()
numbers = [1, 2, 2, 3, 2]

print(numbers.count(2))
```

```python
numbers = [5, 2, 8, 1]

numbers.sort()

print(numbers)

numbers.sort(reverse=True)
print(numbers)
```

```python
# List Concatenation
a = [1, 2]
b = [3, 4]

print(a + b)
```

```python
# List Repetition
a = [1, 2]

print(a * 3)
```

# TUPLES


#### A tuple is similar to a list, but immutable.

```python
# Tuple
# │
# ├── Ordered
# ├── Indexable
# ├── Sliceable
# ├── Immutable
# ├── Allows duplicates
# ├── count()
# ├── index()
# │
# └── Packing / Unpacking
```

```python
# Tuple is Ordered
t = (10, 20, 30)

print(t[0])
```

```python
# Tuple is Immutable

# This doesn't work:

t = (10, 20, 30)

# t[0] = 100

# TypeError
# You cannot change tuple elements.
```

```python
# Why use tuples?
# When data should not be modified.

# Example:
coordinates = (28.61, 77.20)

# You can treat it as a fixed collection.
# Another common example:

rgb = (255, 128, 0)
```

```python
# Tuple with One Element
# This is an interview trap.

# This:
x = (10)

# is not a tuple.
# It's simply:
# int

# For a one-element tuple:

y = (10,)

# The comma is important.
type(y)
type(x)
```

```python
# Tuple Packing
# You can create a tuple without explicitly using parentheses:

person = "Divyanshu", 21, "India"
```

```python
# Tuple Unpacking
# Very useful.

person = ("Divyanshu", 21, "India")

name, age, country = person

print(name)
print(age)
print(country)
```

```python
# Starred Unpacking
numbers = (1, 2, 3, 4, 5)

a, *middle, b = numbers

print(a)
print(middle)
print(b)

# *middle collects the remaining elements into a list.
```

# SETS


What is a Set?
A set is a collection of unique elements.

```python
numbers = {1, 2, 2, 3, 3, 3}

print(numbers)
```

```python
# Sets Don't Allow Duplicates
# This is one of the biggest reasons we use sets.
```

```python
unique = set(numbers)

print(unique)
```

```python
# Sets are Not Indexable

# This doesn't work:

s = {10, 20, 30}

# print(s[0])

TypeError

# Why?
# Because sets don't support positional indexing.
```

```python
# add()
# Adds one element.

s = {1, 2, 3}

s.add(4)

print(s)
```

```python
# update()
# Adds multiple elements.

s = {1, 2}

s.update([3, 4, 5])

print(s)
```

```python
# remove() vs discard()
# Both remove an element.

s = {1, 2, 3}

s.remove(2)
```

```python
# But if element doesn't exist:

s.remove(10)

# → raises KeyError.

# discard() doesn't raise an error:
```

```python
# discard() doesn't raise an error:

s.discard(10)

# Nothing happens.
```

```python
# pop()
# Set has pop(), but there is an important difference from list.

s = {10, 20, 30}

x = s.pop()

# It removes and returns an arbitrary element.
# You should not rely on which element gets removed.
```

```python
# Set Union
# Suppose:

A = {1, 2, 3}
B = {3, 4, 5}
print(A | B)
# Union means:
# Elements present in A OR B.
```

```python
# Set Intersection
print(A & B)

A - B # Set Difference
A ^ B # Symmetric Difference
A.issubset(B) # Subset
B.issuperset(A) # Superset
```

##### Dictionary Keys Must Be Hashable

```python
# Dictionary
# │
# ├── key → value
# ├── Mutable
# ├── Keys are unique
# ├── Keys must be hashable
# ├── Insertion order is preserved
# │
# ├── get()
# ├── keys()
# ├── values()
# ├── items()
# ├── update()
# ├── pop()
# ├── popitem()
# └── clear()
```

```python
d = {
    "name" : "Divyanshu",
    10 : "ten",
    (1, 2): "tuple"
}
```

```python
# But you cannot use a list as a key:

# d = {
#     [1, 2]: "hello"
# }
```

```python
# because lists are mutable and unhashable.

# Simple mental model

# Dictionary needs keys that can be reliably hashed.

# Common hashable types include:

# str
# int
# float
# tuple (if its elements are hashable)

# Lists and sets are not hashable.
```

```python
# Dictionary is Mutable

# You can change values.

student = {
    "name": "Divyanshu",
    "age": 21
}

student["age"] = 22
```
