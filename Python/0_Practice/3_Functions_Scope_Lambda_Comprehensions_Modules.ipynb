````python
# *args
# Suppose you don't know how many positional arguments the function will receive.

def add(*args):
    print(args)

add(10, 20, 30, 40)
# args becomes a tuple: (10, 20, 30, 40)
````

````python
def add(*args):
    total = 0

    for x in args:
        total += x

    return total

print(add(10, 20, 30))
````

#### *args allows a function to accept a variable number of positional arguments. Inside the function, those arguments are available as a tuple.

### The LEGB Rule

#### Python searches for a variable in this order:
1. L → Local
2. E → Enclosing
3. G → Global
4. B → Built-in

````python
y = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print(x)
        print(x + x)
        print(x + y)

    inner()

outer()
````

````python
# If you want to modify the global variable:
x = 10
def change():
    global x
    x = 20

change()

print(x)
````

````python
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()
    print(x)

outer()

# nonlocal means:
# Use the variable from the nearest enclosing function scope.
````

##### global → outside all functions
##### nonlocal → enclosing function

# LAMBDA

##### A lambda is a small anonymous function.

````python
# Normal Function we define
def square(x):
    return x * x
print(square(4))
````

````python
# Syntax
# lambda arguments: expression

square = lambda x: x * x
print(square(5))
````

````python
# Lambda Can Have Multiple Arguments
add = lambda a, b: a + b

print(add(10, 20))
````

````python
# Lambda is useful when you need a small function temporarily, especially as an argument to another function.

# Don't use lambda for complicated logic.
````

````python
# Lambda with sorted()

# This is an important practical example.

# Suppose:

students = [
    ("Aman", 80),
    ("Rahul", 95),
    ("Priya", 70)
]

# Sort according to marks:

students.sort(key=lambda x: x[1])
students
````

````python
# lambda x: x[1]

# means:
# For each tuple, use the second element as the sorting key.
````

# COMPREHENSIONS

##### Python provides a concise way to create collections.

````python
# Normal
numbers = [1, 2, 3, 4, 5]

squares = []

for x in numbers:
    squares.append(x * x)

squares
````

````python
squares1 = [x * x for x in numbers]
squares1
# Same basic result:

# [1, 4, 9, 16, 25]
````

````python
# List Comprehension Structure

# Remember:

# [expression for item in iterable]
# [x * 2 for x in numbers]
````

````python
even = []

for x in numbers:
    if x % 2 == 0:
        even.append(x)
even
````

````python
even1 = [x for x in numbers if x % 2 == 0]

even1
# [expression for item in iterable if condition]
````

````python
# numbers
#  ↓
# filter even
#  ↓
# square
#  ↓
# list
````

````python
numbers = [1, 2, 3, 4]
result = [
    "even" if x % 2 == 0 else "odd"
    for x in numbers
]
result
````

````python
# Set Comprehension

# Same idea:

numbers = [1, 2, 2, 3, 3, 4]

squares = {x * x for x in numbers}

print(squares)
````

````python
numbers = [1, 2, 3, 4]

squares = {x: x * x for x in numbers}
# {key: value for item in iterable}

squares
````

````python
numbers = [1, 2, 3, 4, 5]

even_squares = {
    x: x * x
    for x in numbers
    if x % 2 == 0
}
````

````python
even_squares
````

````python
# Why Use Comprehensions?

# They are useful when the operation is simple.

# Instead of:

result = []

for x in numbers:
    if x > 5:
        result.append(x * 2)

# you can write:

result = [x * 2 for x in numbers if x > 5]

# But don't create extremely complicated comprehensions.
````

```  Why Modules?

````python
# from ... import

# Instead of:

import math

print(math.sqrt(25))

# you can:

from math import sqrt

print(sqrt(25))

# You import only what you need.
````

````python
# Aliases

# You can give a module another name:

import math as m

print(m.sqrt(25))
````

````python
# This is called an alias.

# Common real-world example:

import numpy as np
````

````python
# __name__ == "__main__"

# This is a classic Python interview question.

# Suppose:

def add(a, b):
    return a + b


if __name__ == "__main__":
    print(add(10, 20))

# The block:

# if __name__ == "__main__":

# runs when the file is executed directly.

# If the file is imported as a module, that block doesn't execute.
````

### if __name__ == "__main__": is used to ensure that certain code runs only when the Python file is executed directly, not when the file is imported as a module.

````python
# FUNCTION
# │
# ├── def
# ├── parameters
# ├── arguments
# ├── return
# ├── default arguments
# ├── positional / keyword
# ├── *args → tuple
# └── **kwargs → dictionary


# SCOPE
# │
# └── LEGB
#     ├── Local
#     ├── Enclosing
#     ├── Global
#     └── Built-in


# LAMBDA
# │
# └── lambda arguments: expression


# COMPREHENSION
# │
# ├── [expression for x in iterable]          → List
# ├── {expression for x in iterable}          → Set
# └── {key: value for x in iterable}          → Dictionary


# MODULE
# │
# ├── import module
# ├── from module import thing
# ├── import module as alias
# └── __name__ == "__main__"
````

````python
students = [
    ("Aman", 85),
    ("Rahul", 72),
    ("Priya", 91)
]

result = sorted(
    students,
    key=lambda student: student[1],
    reverse=True # Normal when it is False
)

result

# list
#  ↓
# tuple
#  ↓
# lambda
#  ↓
# sorted()
````
