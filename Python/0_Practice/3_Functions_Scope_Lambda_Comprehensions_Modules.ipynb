#### A parameter is a variable defined in the function declaration, while an argument is the actual value passed to the function when it is called.

```python
def greet(name): # name is paramenter
    print("Hello", name)

greet("Divyanshu") # Divyanshu is argument
```

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

```python
# Caller
#   ↓
# add(10,20)
#   ↓
# function
#   ↓
# return 30
#   ↓
# Caller receives 30
```

```python
# If function uses print():

# x → None

# If function uses return:

# x → 30
```

```python
def calculate(a, b):
    return a + b, a - b

x, y = calculate(10, 5)

print(x)
print(y)
# Internally, Python returns a tuple: (15, 5)
```

```python
def greet(name="User"): # default argument i.e user
    print("Hello", name)

greet()
```

```python
# Argument provided?
#        ↓
#     Yes → use argument
#     No  → use default
```

```python
# *args
# Suppose you don't know how many positional arguments the function will receive.

def add(*args):
    print(args)

add(10, 20, 30, 40)
# args becomes a tuple: (10, 20, 30, 40)
```

```python
def add(*args):
    total = 0

    for x in args:
        total += x

    return total

print(add(10, 20, 30))
```

#### *args allows a function to accept a variable number of positional arguments. Inside the function, those arguments are available as a tuple.

```python
def show(**kwargs):
    print(kwargs)

show(name="Divyanshu", age=21)

# kwargs is a dictionary.
```

#### **kwargs allows a function to accept a variable number of keyword arguments, which are collected into a dictionary.

# Scope


```python
x = 10

def test():
    x = 20
    print(x)

test()
print(x)
```

```python
# Because the x inside the function is a local variable
# Local Scope
# A variable created inside a function generally belongs to that function's local scope.

def test():
    x = 10
    print(x)

test()
```

### The LEGB Rule

#### Python searches for a variable in this order:
1. L → Local
2. E → Enclosing
3. G → Global
4. B → Built-in

```python
y = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print(x)
        print(x + x)
        print(x + y)

    inner()

outer()
```

```python
# If you want to modify the global variable:
x = 10
def change():
    global x
    x = 20

change()

print(x)
```

```python
def outer():
    x = 10

    def inner():
        nonlocal x
        x = 20

    inner()
    print(x)

outer()

# nonlocal means:
# Use the variable from the nearest enclosing function scope.
```
