```python
s = input()

temp = ""

for ch in  s:
    if ch.isalnum():
        temp += ch.lower()

ans = True

print(temp)

i = 0
j = len(temp) - 1

while i < j:
    if temp[i] != temp[j]:
        ans = False
    
    i += 1
    j -= 1

print(ans)

```

```python
def clean_node(s):
    parts = s.split(",")

    result = []

    for part in parts:
        part = part.strip()
        part = part.lower()

        if part != "":
            result.append(part)
    return ",".join(result)

s = " Node-A, node-b ,NODE->C, , node-d, "
print(clean_node(s))
```

```python
def is_valid(s):
    stack = []

    pairs = {
        ')': '(',
        ']': '[',
        '}': '{',
    }

    for ch in s:

        # Openening bracket
        if ch in "[({":
            stack.append(ch)

        # Closing bracket 
        else:
            # No opening bracket available
            if not stack:
                return False
            
            #Top opening bracket doesn't match
            if stack[-1] != pairs[ch]:
                return False
        
        stack.pop()

    # Stack should be empty at the end
    return len(stack) == 0

print(is_valid("()[]{}"))   # True
print(is_valid("([{}])"))   # True
print(is_valid("(]"))       # False
print(is_valid("([)]"))      # False
print(is_valid("((("))      # False
print(is_valid("]"))        # False
```
