#### Problem: Write a function that takes a list of numbers and returns a list (or set) of all the elements that appear more than once (duplicates). For example, for [4, 3, 2, 7, 8, 2, 3, 1], the duplicates are [2, 3].

```python
def duplicates(arr: list):
    temp = {}
    for num in arr:
        temp[num] = temp.get(num, 0) + 1

    ans = []
    for num, count in temp.items():
        if count > 1:
            ans.append(num)
    return ans

arr = [4, 3, 2, 7, 8, 2, 3, 1]
print(duplicates(arr))
```

#### Problem: Given an array of integers nums and an integer target, return the indices of the two numbers such that they add up to target. You may assume that each input would have exactly one solution, and you cannot use the same element twice. You can return the answer in any order.

<!-- Example:

Input: nums = [2, 7, 11, 15], target = 9

Output: [0, 1] (Because nums[0] + nums[1] == 2 + 7 == 9) -->

```python
def two_sum(nums, target):
    temp = {}
    ans = []
    for indx, num in  enumerate(nums):
        if target - num in temp:
            ans.append(temp[target-num])
            ans.append(indx)
        
        temp[num] = temp.get(num, 0) + indx

    return ans

nums = [2, 7, 11, 15]
target = 13

print(two_sum(nums, target))
```

#### Problem: Given an integer array nums, move all 0's to the end of it while maintaining the relative order of the non-zero elements. (Note: You must do this in-place without making a copy of the array).

Example:

Input: nums = [0, 1, 0, 3, 12]

Output: [1, 3, 12, 0, 0]

```python
def move_all_zeroes(nums):
    i = 0
    j = 0

    while j < len(nums):
        if nums[j] != 0:
            nums[i], nums[j] = nums[j], nums[i]
            i += 1
        j += 1
    
nums = [0, 1, 0, 3, 12]
move_all_zeroes(nums)
print(nums)
        

```

#### Find Maximum and Minimum Without Built-in Functions
Problem: Write a function that takes a list of numbers and finds both the maximum and minimum elements without using built-in functions like max() or min(). Return the result as a tuple (max_val, min_val).

Example:

Input: [5, 2, 9, 1, 7, 6]

Output: (9, 1)

```python
def max_min(nums: list)->tuple:
    mx = nums[0]
    mn = nums[0]

    for num in nums:
        if num > mx:
            mx = num

        if mn > num:
            mn = num
    
    return (mx, mn)

nums = [5, 2, 9, 1, 7, 6]
print(max_min(nums))
```

#### Problem: Given two sorted lists (or arrays) list1 and list2, merge them into a single sorted list. (Do this efficiently without just doing list1 + list2 and sorting it again, try using a two-pointer approach).

Example:

Input: list1 = [1, 3, 5], list2 = [2, 4, 6]

Output: [1, 2, 3, 4, 5, 6]

```python
def merge_two_sorted_lists(l1, l2)->list:
    ans = []
    i = 0
    j = 0

    while i < len(l1) and j < len(l2):
        if l1[i] <= l2[j]:
            ans.append(l1[i])
            i += 1
        else:
            ans.append(l2[j])
            j += 1

    while i < len(l1):
        ans.append(l1[i])
        i += 1

    while j < len(l2):
        ans.append(l2[j])
        j += 1
    return ans

list1 = [1, 3, 5]
list2 = [2, 4, 6]

print(merge_two_sorted_lists(list1, list2))
```
