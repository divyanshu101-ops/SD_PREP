##### Problem: Given a string, write a Python function/code to reverse it. (Try writing it without using direct slicing like [::-1] first if you want to practice logic, though slicing bhi valid Pythonic approach hai—tu apne hisaab se try kar!).

```python
s = list("Divyanshu Gautam") # Create list using string

i = 0
j = len(s) - 1

while i < j:
    temp  = s[i]
    s[i] = s[j]
    s[j] = temp

    i += 1
    j -= 1

s = "".join(s)
print(s)
```

```python
def reverse_string(s: str) -> str:
    # String immutable hoti hai, isliye list me convert karna padega
    chars = list(s)
    i = 0
    j = len(chars) - 1
    print(chars)
    while i < j:
        chars[i], chars[j] = chars[j], chars[i]  # Pythonic swap (temp variable ki bhi zaroorat nahi!)
        i += 1
        j -= 1
        
    return "".join(chars)

# Test
s = "Divyanshu Gautam"
print(reverse_string(s))  # Output: 'mutuaG uhsnayviD'
```

#### Write a function that checks whether a given string is a palindrome (reads the same forward and backward, e.g., "madam" or "racecar"). The function should return True or False.

```python
def is_palindrome(s: str) -> bool:
    chars = list(s)
    i = 0
    j = len(chars)-1
    flag = True

    while i < j:
        if chars[i] != chars[j]:
            flag = False
        i += 1
        j -= 1
    
    return flag

s = "madam"
print(is_palindrome(s))

s1 = "rabecar"
print(is_palindrome(s1))
```

```python
def is_palindrome(s: str) -> bool:
    i = 0
    j = len(s) - 1

    while i < j:
        if s[i] != s[j]:
            return False  # Yahin se turant False bhej diya (Short-circuiting)
        i += 1
        j -= 1
        
    return True

print(is_palindrome("madam"))   # True
print(is_palindrome("rabecar")) # False
```

#### Write a function that takes two strings and checks if they are anagrams of each other (i.e., they contain the exact same characters with the same frequencies, just in a different order, like "listen" and "silent"). Return True or False.

```python

def is_anagram(s1: str, s2: str) -> bool:
    if len(s1) != len(s2):
        return False
        
    freq1 = {}
    freq2 = {}
    for ch in s1:
        freq1[ch] = freq1.get(ch, 0) + 1

    for ch in s2:
        freq2[ch] = freq2.get(ch, 0) + 1

    for ch in freq1:
        if freq1[ch] != freq2[ch]:
            return False

    return True

s1 = "listen" 
s2 = "silnt"

print(is_anagram(s1, s2))
```

#### Problem: Write a function that takes a string and returns the count of vowels (a, e, i, o, u) and consonants in it. (You can ignore spaces or handle them gracefully).

```python
def numberOf_consonant_vowels(s : str) -> tuple:
    l = ['a', 'e', 'i', 'o', 'u']
    countConsonant = 0
    countVowels = 0

    for ch in s:
        if ch == ' ':
            continue
        elif ch.lower() in l:
            countVowels += 1
        else:
            countConsonant += 1

    return (countVowels, countConsonant)

t = numberOf_consonant_vowels("Divyashu gautam")
print("Vowels:", t[0])
print("Consonant:", t[1])
```

```python
def numberOf_consonant_vowels(s: str) -> tuple:
    # Set use karna 'in' check ke liye zyada fast hota hai
    vowels = {'a', 'e', 'i', 'o', 'u'}
    count_consonant = 0
    count_vowels = 0

    for ch in s.lower():
        if ch == ' ':  # Space ko skip karne ke liye
            continue
        
        # Check karte hain ki character alphabetic hai ya nahi
        if ch.isalpha():
            if ch in vowels:
                count_vowels += 1
            else:
                count_consonant += 1

    return (count_vowels, count_consonant)  # Tuple return kiya () ke sath

t = numberOf_consonant_vowels("Divyanshu gautam")
print("Vowels:", t[0])
print("Consonants:", t[1])
```

#### Problem: Write a function that takes a string and finds the index (or character) of the first non-repeating character in it. If all characters repeat, return -1 or an appropriate indicator (e.g., for "leetcode" -> 'l', because 'l' is the first character that doesn't repeat).

```python
def non_repeating_character(s: str)-> int:
    freq = {}

    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    ans = ''
    for ch in  freq:
        if freq.get(ch) == 1:
            ans = ch
            break
    
    if ans != '':
        return s.find(ans)

    return -1

num = non_repeating_character("leettcodel")
print(num)
```

```python
def non_repeating_character(s: str) -> int:
    freq = {}

    # Step 1: Frequency count karo
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    # Step 2: String par dobara iterate karke pehla character dhundo jinki frequency 1 ho
    for i in range(len(s)):
        if freq[s[i]] == 1:
            return i  # Turant index return kar diya

    return -1  # Agar koi non-repeating character na ho

num = non_repeating_character("leettcodel")
print(num)  # Output corresponding to 'c' index
```

#### Given a string s, find the length of the longest substring without repeating characters (e.g., for "abcabcbb", the answer is 3 because the longest substring without repeating characters is "abc").

```python
def length_of_long_substring(s: str) -> int:
    char_map = {}  # Character aur uska last seen index store karne ke liye
    max_len = 0
    i = 0  # Left pointer of the window
    
    for j in range(len(s)):
        ch = s[j]
        
        # Agar character map me hai aur uska index current window (i) ke andar ya uske baad hai
        if ch in char_map and char_map[ch] >= i:
            # Left pointer ko duplicate character ke theek aage shift kar do
            i = char_map[ch] + 1
            
        # Current character ka latest index update karo
        char_map[ch] = j
        
        # Max length update karo (current window size = j - i + 1)
        max_len = max(max_len, j - i + 1)
        
    return max_len

# Test
s = "abcabcbb"
print(length_of_long_substring(s))  # Output: 3 ("abc")
```
